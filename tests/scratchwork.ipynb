```python
from apex.metrics import compute_segment_metrics
from apex.parsing import GarminParser, RollSegmenter
from apex.parsing.schema import CourseSpec, RollSegment
```

```python
with open("../src/apex/parsing/course_spec.json") as f:
    course_spec = CourseSpec.model_validate_json(f.read())
```

```python
from pathlib import Path

# file_path = "sample_data/GA373120.FIT"
data_dir = Path("/home/nobe/dev/apex/GPS-Save-Our-Drivers-mod/Rolls Data")
paths = list(data_dir.glob("*.FIT"))

file_path = str(paths[50])
print(file_path)

def parse_roll_from_fit(file_path: str) -> dict[str, RollSegment]:
    parser = GarminParser()
    segmenter = RollSegmenter(course_spec)

    roll_data = parser.parse_file(file_path)
    segments = segmenter.segment(roll_data)
    return segments
```

```python
for path in paths:
    result = parse_roll_from_fit(str(path))
```

```python
import numpy as np

metrics = compute_segment_metrics(result["monument_straight"])

np.diff(metrics.seg_dist_m) / np.diff(metrics.seg_time_sec)
np.gradient(metrics.seg_dist_m, metrics.seg_time_sec)
```
